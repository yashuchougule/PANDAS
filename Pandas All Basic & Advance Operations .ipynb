```python
# Column Transformations In Pandas (Part1) = 
path2 = pd.read_csv("C:/Users/DSi/Desktop/student_data_A.csv")
print(path2)
```

```python
path2.loc[(path2["Percentage"] == 35), "Result"] = "Pass"
path2.loc[(path2["Percentage"] > 50), "Result"] = "Third Class"
path2.loc[(path2["Percentage"] > 65), "Result"] = "Second Class"
path2.loc[(path2["Percentage"] > 75), "Result"] = "First Class"
path2.loc[(path2["Percentage"] > 85), "Result"] = "Distinction"
print(path2)
```

```python
path2["Full Name"] = path2["First Name"] +" "+ path2["Last Name"]
print(path2)
```
